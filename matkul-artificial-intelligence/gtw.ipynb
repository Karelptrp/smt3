```python
import numpy as np

X = np.array([[18.0, 250.0],
             [25.0, 400.0],
             [42.0, 800.0]])
X_min = np.min(X, axis=0)
X_max = np.max(X, axis=0)
X_scaled = (X - X_min) / (X_max - X_min)
print("Hasil Scaling:\n", X_scaled)
```
